```python
# Import necessary libraries
import pandas as pd
import numpy as np

# Load the dataset
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv'
df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
```

```python
# Part 1: Initial Data Exploration
# Basic Dataset Information

# 1. First 5 and last 4 rows
print(" - First 5 rows - ")
print(df.head())
print(" - Last 5 rows - ")
print(df.tail())

# 2. Data type of the dataset
print("\n - Dataset type - ")
print(type(df))

# 3. Shape
print("\n - Shape (rows, columns) - ")
print(df.shape)

# 4. Total size
print("\n - Total size (elements) - ")
print(df.size)

# 5. Column names and dtypes
print(" - Column names - ")
print(df.columns.tolist())
print("\n - Column Data types - ")
print(df.dtypes)
```

```python
# Data Structure Analysis

# 1. Detailed info
print(" - Dataframe Info - ")
df.info()

# 2. Index
print("\n - Index - ")
print(df.index)
print("Index type:", type(df.index))
print("Index represents: the default row number (0-based) for each record")

# 3. Count columns - two methods
print("\n - Column count - ")
print("Method 1 (len(df.columns)):", len(df.columns))
print("Method 2 (df.shape[1])", df.shape[1])

# 4. Count rows - two methods
print("\n - Row count - ")
print("Method 1 (len(df)):", len(df))
print("Method 2 (df.shape[0]):", df.shape[0])

# 5. Numerical vs. categorical columns
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
print("\n - Numerical columns - ")
print(numerical_cols)
print("\n - Categorical columns - ")
print(categorical_cols)
```

```python
# Missing Data Investigation

# 1. Any missing values?
print(" - Any missing values - ")
print(df.isnull().any())
print("Any missing at all:", df.isnull().any().any())

# 2. Count missing per column
print("\n - Missing value count per column - ")
print(df.isnull().sum())

# 3. Percentage missing per column
print("\n - Missing value percentage per column - ")
missing_pct = (df.isnull().sum() / len(df)) * 100
print(missing_pct)

# 4. Column with most missing data
print("\n - Columns with most missing data - ")
print(missing_pct.sort_values(ascending=False).head())
```

```python
# Part 2: Descriptive Statistics Analysis
# Numerical Data Analysis

# 1. Describe all numerical columns
print(" - Describe numerical columns - ")
print(df.describe())

# 2. Total bill detailed stats
print("\n - Total Bill Statistics - ")
print("Mean:", df['total_bill'].mean())
print("Median:", df['total_bill'].median())
print("Mode:", df['total_bill'].mode()[0])
print("Std Dev:", df['total_bill'].std())
print("Variance:", df['total_bill'].var())
print("Min:", df['total_bill'].min())
print("Max:", df['total_bill'].max())
print("25th percentile:", df['total_bill'].quantile(0.25))
print("75th percentile:", df['total_bill'].quantile(0.75))

# 3. Highest total bill
print("\n - Highest Total Bill - ")
print(df['total_bill'].max())
print(df.loc[df['total_bill'].idxmax()])

# 4. Lowest total bill
print("\n - Lowest Total Bill - ")
print(df['total_bill'].min())
print(df.loc[df['total_bill'].idxmin()])

# 5. Bills above average
avg_bill = df['total_bill'].mean()
above_avg = (df['total_bill'] > avg_bill).sum()
print(f"\n - Bills Above Average (${avg_bill:.2f}) - ")
print(f"Count: {above_avg} out of {len(df)} ({above_avg/len(df)*100:.1f}%)")
```

```python
# Categorical Data Analysis

# 1. Describe categorical columns
print(" - Describe Categorical columns - ")
print(df.describe(include=['object']))

# 2. For each categorical column
print("\n - Per Categorical column analysis - ")
for col in df.select_dtypes(include='object').columns:
  print("Unique values:", df[col].nunique())
  print("Most frequent values:", df[col].mode()[0])
  print("Value counts:\n", df[col].value_counts())
  print("Frequency of most common value:", df[col].value_counts().iloc[0])

# 3. Note: This dataset has no region/continent column.
print("\n - Note: No region/continent column exists in the Tips dataset - ")
print("Distribution by 'day' (closest analog to a regional grouping):")
print(df['day'].value_counts())
```

```python
# Adavanced Statistical Insights

# 1. Correlation between total_bill and tip
print(" - Correlation: total_bill vs tip - ")
print("Pearson correlation:", df['total_bill'].corr(df['tip']))
print("\nFull correlation matrix:")
print(df[['total_bill', 'tip', 'size']].corr())

# 2. Top 5 Highest tips
print("\n - Top 5 Highest tips - ")
print(df.nlargest(5, 'tip')[['total_bill', 'tip', 'sex', 'smoker', 'day', 'time', 'size']])

# 3. lowest tips
print("\n - 5 Lowest tips - ")
print(df.nsmallest(5, 'tip')[['total_bill', 'tip', 'sex', 'smoker', 'day', 'time', 'size']])

# 4. Range for bill and tip
print("\n - Ranges - ")
print(f"Bill range: {df['total_bill'].max() - df['total_bill'].min():.2f}")
print(f"Tip range: {df['tip'].max() - df['tip'].min():.2f}")

# 5. % of tips above $4.00
pct_above_4 = (df['tip'] > 4.00).mean() * 100
print(f"\n - Tips above $4.00: {pct_above_4:.2f}% - ")
print(f"Count: {(df['tip'] > 4.00).sum()} out of {len(df)}")

```

```python
# Part 3: Data Quality Assessment
# Data Consistency Check

# 1. Duplicate rows
print(" - Duplicate Rows - ")
print("Number of duplicate rows:", df.duplicated().sum())

# 2. Positive values check
print("\n Positive value check - ")
print("Any bill <= 0:", (df['total_bill'] <= 0).any())
print("Any tip <= 0:", (df['tip'] <=0).any())

# 3. Outlier check via IQR for total_bill
Q1 = df['total_bill'].quantile(0.25)
Q3 = df['total_bill'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df['total_bill'] < lower)| (df['total_bill'] > upper)]
print("\n - Outliers in total_bill (IQR method) - ")
print(f"Bounds: [{lower:.2f}, {upper:.2f}]")
print(f"Number of outliers: {len(outliers)}")
print(outliers[['total_bill', 'tip', 'day', 'time']])

# 4. Consistent formatting
print("\n -  Categorical formatting check - ")
for col in ['sex', 'smoker', 'day', 'time']:
  print(f"{col}: {sorted(df[col].unique())}")
```

```python
# Data Distribution Analysis

# 1. Quartiles for tip
print(" - Tip Quartiles - ")
print("Q1 (25%):", df['tip'].quantile(0.25))
print("Q2 (50%/median):", df['tip'].quantile(0.50))
print("Q3 (75%):", df['tip'].quantile(0.75))


# 2. Outliers in tip
Q1t = df['tip'].quantile(0.25)
Q3t = df['tip'].quantile(0.75)
IQRt = Q3t -Q1t
lo, hi = Q1t - 1.5*IQRt, Q3t + 1.5*IQRt
tip_outliers = df[(df['tip'] < lo) | (df['tip'] > hi)]
print(f"\n - Tip Outliers - Bounds [{lo:.2f}, {hi:.2f}], count: {len(tip_outliers)}")

# 3. Median split
median_tip = df['tip'].median()
above = (df['tip'] > median_tip).sum()
below = (df['tip'] < median_tip).sum()
equal = (df['tip'] == median_tip).sum()
print(f"\n - Tip Median split (median = ${median_tip}) - ")
print(f"Above: {above}, Below: {below}, Equal: {equal}")

# 4. Value counts for categoricals
print("\n - Value counts - ")
for col in ['sex', 'smoker', 'day', 'time']:
  print(f"\n - {col} - ")
  print(df[col].value_counts())
```

# Part 4: Summary and Insights

## Summary Report - Tips Dataset

### 1. Dataset Overview

This dataset can be thought of as a collection of 244 restaurant receiptsm each recording the total bill, the tip left, and some basic information about the customer and the dining context, including sex, smoking status, day of the week, time of day(lunch or dinner), and party size.

The **Tips dataset** contains **244 restaurant bills** with **7 features** per bill: three numerical columns (bill amount, tip amount, party size) and four categorical columns (sex, smoker, day, time). A notable strength of this dataset is its cleanliness: there are **no missing values anywhere**, no negative amounts, and only one duplicate row, meaning no imputation or heavy cleaning is needed before analysis.

### 2. Key Findings

- Average tip is $3.00 and average bill is $19.79.
- Median tip is $2.99 and median bill is $17.80, showing a slight right skew (a few larg bills pull the mean up).
- Highest tip was $10.00 on a $50.81 bill during Saturday dinner.
- Lowest tip was $1.00 on a $3.07 bill during Saturday dinner.
- Bill amount and tip amount have a strong positive correlation of 0.676, meaning larger bills tend to receive larger tips.
- 99 bills (40.6%) were above the average bill amount.
- 41 tips (16.8%) were above $4.00.
- Saturday was the busiest day (87 bills) and Friday was the quietest (19 bills).
- Dinner (176 bills) was far more common than lunch (68 bills).
- Male customers (157) outnumbered female customers (87).
- Non-smokers (151) outnumbered smokers (93).
- **Most important insight:** Tip amounts scale strongly with bill size, but tip *percentages* vary widely across bills — suggesting that tipping is partly proportional and partly driven by habit or context.

### 3. Data Quality Report

**Missing data:** There are no missing values in any column, meaning every record is complete and the dataset is ready for analysis and imputation.

**Duplicates:** One duplicate row was detected. It may represent either a geniune repeat visit or a data-entry error and should be reviewed.

**Outliers:** Using the IQR method, 9 unusually high bills were identified (values above $40.30) and 9 unusually large tips were flagged (values above $5.91). These appear to be valid real-world values. Likely large parties rather than erorrs.

**A note on the median split:** The tip median is $2.90, and the 244 records split evenly with 122 above and 122 below the median. No records equal $2.90 exactly, which is a floating-point representation quirk (the value 2.90 cannot be stored exactly as a binary float). The distribution is nonetheless well-centered.

**Consistency:** All categorical values are cleanly formatted with no stray typos or capitalization mismatches (Male, Female, No/Yes, Fri/Sat/Sun/Thur, Lunch/Dinner), which makes the data easy to group and analyze.

**Recommendation for improvements:**
1. Investigate and remove the single duplicate row if confirmed as a data-entry error.
2. Add a derived 'tip_percentage' column for richer proportional analysis.
3. Collect contextual variables such as service quality and part composition to help explain the wide variance in tipping behavior.
